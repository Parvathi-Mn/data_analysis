```python
import os
import json
import glob
import numpy as np
from datetime import datetime

dataset_path = '/content/drive/MyDrive/Behaviour Biometrics Dataset/raw_kmt_dataset/'
json_files = glob.glob(os.path.join(dataset_path, "*.json"))

data_by_user = {}

print(f"Parsing {len(json_files)} user files with datetime handling...")

for file_path in json_files:
    filename = os.path.basename(file_path)
    user_id = "user_" + filename.split('_')[3].split('.')[0]

    with open(file_path, 'r') as f:
        content = json.load(f)

    user_sequences = []
    true_data = content.get('true_data', {})

    for test_key, test_session in true_data.items():
        key_events = test_session.get('key_events', [])
        feature_rows = []

        press_times = {}
        last_release_time = None
        session_start_dt = None

        for event in key_events:
            key_name = event.get('Key', '')
            event_type = event.get('Event', '') # 'pressed' or 'released'
            time_str = event.get('Timestamp', event.get('time', ''))

            try:
                # Parse timestamp string to datetime object
                dt = datetime.strptime(time_str.split('+')[0].strip(), '%Y-%m-%d %H:%M:%S.%f')
            except ValueError:
                try:
                    # Fallback format if microseconds are missing
                    dt = datetime.strptime(time_str.split('+')[0].strip(), '%Y-%m-%d %H:%M:%S')
                except:
                    continue

            if session_start_dt is None:
                session_start_dt = dt

            # Convert datetime to elapsed seconds from session start
            timestamp = (dt - session_start_dt).total_seconds()

            dwell_time = 0.0
            flight_time = 0.0

            if event_type == 'pressed':
                press_times[key_name] = timestamp
                if last_release_time is not None:
                    flight_time = max(0.0, timestamp - last_release_time)
            elif event_type == 'released' and key_name in press_times:
                dwell_time = max(0.0, timestamp - press_times[key_name])
                last_release_time = timestamp
                del press_times[key_name]

            # Simple feature representation per event: [dwell_time, flight_time, timestamp_delta]
            feat_vector = [
                dwell_time,
                flight_time,
                timestamp
            ]
            feature_rows.append(feat_vector)

        # Chunk into fixed-length windows (e.g., 30 events per window)
        if len(feature_rows) >= 30:
            seq_len = 30
            feature_array = np.array(feature_rows, dtype=np.float32)

            for i in range(0, len(feature_array) - seq_len + 1, seq_len // 2):
                window = feature_array[i:i+seq_len]
                user_sequences.append(window)

    if len(user_sequences) > 0:
        data_by_user[user_id] = user_sequences

print(f"Successfully parsed data for {len(data_by_user)} users!")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import numpy as np
import random
from sklearn.metrics import roc_curve, auc
from scipy.optimize import brentq
from scipy.interpolate import interp1d

# ==========================================
# 1. TRIPLET DATASET WRAPPER
# ==========================================
class BehavioralDataset(Dataset):
    def __init__(self, data_by_user):
        # Filter out users with too few samples to form triplets
        self.data_by_user = {u: seqs for u, seqs in data_by_user.items() if len(seqs) >= 2}
        self.users = list(self.data_by_user.keys())

    def __len__(self):
        return sum(len(samples) for samples in self.data_by_user.values())

    def __getitem__(self, idx):
        anchor_user = random.choice(self.users)
        samples = self.data_by_user[anchor_user]
        anchor_seq, positive_seq = random.sample(samples, 2)

        negative_user = random.choice([u for u in self.users if u != anchor_user])
        negative_seq = random.choice(self.data_by_user[negative_user])

        return (
            torch.tensor(anchor_seq, dtype=torch.float32),
            torch.tensor(positive_seq, dtype=torch.float32),
            torch.tensor(negative_seq, dtype=torch.float32)
        )

# ==========================================
# 2. MODEL DEFINITIONS
# ==========================================

# Model 1: Siamese CNN + BiGRU (Main Candidate)
class CNNBiGRUEmbedding(nn.Module):
    def __init__(self, input_dim, hidden_dim=128, embedding_dim=64):
        super().__init__()
        self.conv1 = nn.Conv1d(in_channels=input_dim, out_channels=64, kernel_size=3, padding=1)
        self.relu = nn.ReLU()
        self.pool = nn.MaxPool1d(kernel_size=2)
        self.bigru = nn.GRU(input_size=64, hidden_size=hidden_dim, num_layers=2, batch_first=True, bidirectional=True)
        self.fc = nn.Linear(hidden_dim * 2, embedding_dim)

    def forward_once(self, x):
        x = x.permute(0, 2, 1) # (batch, input_dim, seq_len)
        x = self.pool(self.relu(self.conv1(x)))
        x = x.permute(0, 2, 1) # (batch, seq_len, channels)
        out, _ = self.bigru(x)
        x = torch.mean(out, dim=1)
        return nn.functional.normalize(self.fc(x), p=2, dim=1)

    def forward(self, anchor, positive, negative):
        return self.forward_once(anchor), self.forward_once(positive), self.forward_once(negative)

# Model 2: Siamese Transformer Encoder
class TransformerEmbedding(nn.Module):
    def __init__(self, input_dim, model_dim=64, num_heads=4, num_layers=2, embedding_dim=64):
        super().__init__()
        self.input_proj = nn.Linear(input_dim, model_dim)
        encoder_layer = nn.TransformerEncoderLayer(d_model=model_dim, nhead=num_heads, dim_feedforward=128, batch_first=True)
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)
        self.fc = nn.Linear(model_dim, embedding_dim)

    def forward_once(self, x):
        x = self.input_proj(x)
        x = self.transformer(x)
        x = torch.mean(x, dim=1)
        return nn.functional.normalize(self.fc(x), p=2, dim=1)

    def forward(self, anchor, positive, negative):
        return self.forward_once(anchor), self.forward_once(positive), self.forward_once(negative)

# Model 3: Siamese TCN
class TemporalBlock(nn.Module):
    def __init__(self, n_inputs, n_outputs, kernel_size, stride, dilation, padding):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv1d(n_inputs, n_outputs, kernel_size, stride=stride, padding=padding, dilation=dilation),
            nn.ReLU(),
            nn.Conv1d(n_outputs, n_outputs, kernel_size, stride=stride, padding=padding, dilation=dilation),
            nn.ReLU()
        )
        self.downsample = nn.Conv1d(n_inputs, n_outputs, 1) if n_inputs != n_outputs else None

    def forward(self, x):
        out = self.net(x)
        res = x if self.downsample is None else self.downsample(x)
        return nn.functional.relu(out + res)

class TCNEmbedding(nn.Module):
    def __init__(self, input_dim, num_channels=[64, 64], kernel_size=3, embedding_dim=64):
        super().__init__()
        layers = []
        for i, out_channels in enumerate(num_channels):
            dilation_size = 2 ** i
            in_channels = input_dim if i == 0 else num_channels[i-1]
            layers.append(TemporalBlock(in_channels, out_channels, kernel_size, stride=1, dilation=dilation_size, padding=(kernel_size-1)*dilation_size))
        self.tcn = nn.Sequential(*layers)
        self.fc = nn.Linear(num_channels[-1], embedding_dim)

    def forward_once(self, x):
        x = x.permute(0, 2, 1) # (batch, channels, seq_len)
        x = self.tcn(x)
        x = torch.mean(x, dim=2)
        return nn.functional.normalize(self.fc(x), p=2, dim=1)

    def forward(self, anchor, positive, negative):
        return self.forward_once(anchor), self.forward_once(positive), self.forward_once(negative)

# ==========================================
# 3. STANDARDIZED EVALUATOR
# ==========================================
class BiometricEvaluator:
    def __init__(self, model, device):
        self.model = model
        self.device = device
        self.model.to(device)
        self.model.eval()

    def get_embeddings(self, data_dict):
        user_embeddings = {}
        with torch.no_grad():
            for user_id, sequences in data_dict.items():
                if len(sequences) == 0:
                    continue
                tensor_seqs = torch.tensor(np.array(sequences), dtype=torch.float32).to(self.device)
                embeddings = self.model.forward_once(tensor_seqs).cpu().numpy()
                user_embeddings[user_id] = embeddings
        return user_embeddings

    def evaluate_trials(self, user_embeddings, k_shots):
        genuine_scores, impostor_scores = [], []
        user_ids = list(user_embeddings.keys())

        for uid in user_ids:
            all_seqs = user_embeddings[uid]
            if len(all_seqs) <= k_shots:
                continue

            profile = np.mean(all_seqs[:k_shots], axis=0)
            profile = profile / np.linalg.norm(profile)

            for probe in all_seqs[k_shots:]:
                genuine_scores.append(1.0 - np.dot(profile, probe))

            for other_uid in user_ids:
                if other_uid == uid:
                    continue
                for probe in user_embeddings[other_uid]:
                    impostor_scores.append(1.0 - np.dot(profile, probe))

        return np.array(genuine_scores), np.array(impostor_scores)

    def compute_metrics(self, genuine_scores, impostor_scores):
        if len(genuine_scores) == 0 or len(impostor_scores) == 0:
            return {"EER": 0.0, "ROC-AUC": 0.0, "TAR@1%FAR": 0.0}

        y_true = np.concatenate([np.zeros(len(genuine_scores)), np.ones(len(impostor_scores))])
        y_scores = np.concatenate([genuine_scores, impostor_scores])

        fpr, tpr, thresholds = roc_curve(y_true, -y_scores)
        roc_auc = auc(fpr, tpr)

        try:
            eer = brentq(lambda x: 1.0 - x - interp1d(fpr, tpr)(x), 0.0, 1.0)
        except:
            eer = 0.0

        try:
            tar_at_far = float(interp1d(fpr, tpr, fill_value="extrapolate")(0.01))
        except:
            tar_at_far = 0.0

        return {"EER": float(eer), "ROC-AUC": float(roc_auc), "TAR@1%FAR": tar_at_far}

# ==========================================
# 4. EXECUTION & TRAINING PIPELINE
# ==========================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
input_dim = 3 # [dwell_time, flight_time, timestamp_delta]

dataset = BehavioralDataset(data_by_user)
dataloader = DataLoader(dataset, batch_size=32, shuffle=True)

models = {
    "CNN-BiGRU": CNNBiGRUEmbedding(input_dim=input_dim).to(device),
    "Transformer": TransformerEmbedding(input_dim=input_dim).to(device),
    "TCN": TCNEmbedding(input_dim=input_dim).to(device)
}

shot_counts = [1, 3, 5, 10]
epochs = 5 # Increase to 15-20 later for full convergence

for name, model in models.items():
    print(f"\n=========================================")
    print(f"Training Model: {name}")
    print(f"=========================================")

    optimizer = optim.Adam(model.parameters(), lr=0.001)
    criterion = nn.TripletMarginLoss(margin=0.5, p=2)

    model.train()
    for epoch in range(epochs):
        total_loss = 0.0
        for anchor, positive, negative in dataloader:
            anchor, positive, negative = anchor.to(device), positive.to(device), negative.to(device)
            optimizer.zero_grad()
            anc_emb, pos_emb, neg_emb = model(anchor, positive, negative)
            loss = criterion(anc_emb, pos_emb, neg_emb)
            loss.backward()
            optimizer.step()
            total_loss += loss.item()
        print(f"Epoch [{epoch+1}/{epochs}], Loss: {total_loss / len(dataloader):.4f}")

    print(f"\nEvaluating {name} Across Few-Shot Protocols:")
    evaluator = BiometricEvaluator(model, device)
    user_embeddings = evaluator.get_embeddings(data_by_user)

    for k in shot_counts:
        gen_scores, imp_scores = evaluator.evaluate_trials(user_embeddings, k_shots=k)
        metrics = evaluator.compute_metrics(gen_scores, imp_scores)
        print(f"  -> {k}-shot | EER: {metrics['EER']:.4f} | AUC: {metrics['ROC-AUC']:.4f} | TAR@1%FAR: {metrics['TAR@1%FAR']:.4f}")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import numpy as np
import random
from sklearn.metrics import roc_curve, auc
from scipy.optimize import brentq
from scipy.interpolate import interp1d

# ==========================================
# 1. TRIPLET DATASET WRAPPER
# ==========================================
class BehavioralDataset(Dataset):
    def __init__(self, data_by_user):
        self.data_by_user = {u: seqs for u, seqs in data_by_user.items() if len(seqs) >= 2}
        self.users = list(self.data_by_user.keys())

    def __len__(self):
        return sum(len(samples) for samples in self.data_by_user.values())

    def __getitem__(self, idx):
        anchor_user = random.choice(self.users)
        samples = self.data_by_user[anchor_user]
        anchor_seq, positive_seq = random.sample(samples, 2)

        negative_user = random.choice([u for u in self.users if u != anchor_user])
        negative_seq = random.choice(self.data_by_user[negative_user])

        return (
            torch.tensor(anchor_seq, dtype=torch.float32),
            torch.tensor(positive_seq, dtype=torch.float32),
            torch.tensor(negative_seq, dtype=torch.float32)
        )

# ==========================================
# 2. SIAMESE TRANSFORMER ARCHITECTURE
# ==========================================
class TransformerEmbedding(nn.Module):
    def __init__(self, input_dim, model_dim=64, num_heads=4, num_layers=2, embedding_dim=64):
        super().__init__()
        self.input_proj = nn.Linear(input_dim, model_dim)
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=model_dim,
            nhead=num_heads,
            dim_feedforward=128,
            batch_first=True,
            dropout=0.3
        )
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)
        self.fc = nn.Linear(model_dim, embedding_dim)

    def forward_once(self, x):
        x = self.input_proj(x)
        x = self.transformer(x)
        x = torch.mean(x, dim=1)
        return nn.functional.normalize(self.fc(x), p=2, dim=1)

    def forward(self, anchor, positive, negative):
        return self.forward_once(anchor), self.forward_once(positive), self.forward_once(negative)

# ==========================================
# 3. STANDARDIZED EVALUATOR
# ==========================================
class BiometricEvaluator:
    def __init__(self, model, device):
        self.model = model
        self.device = device
        self.model.to(device)
        self.model.eval()

    def get_embeddings(self, data_dict):
        user_embeddings = {}
        with torch.no_grad():
            for user_id, sequences in data_dict.items():
                if len(sequences) == 0:
                    continue
                tensor_seqs = torch.tensor(np.array(sequences), dtype=torch.float32).to(self.device)
                embeddings = self.model.forward_once(tensor_seqs).cpu().numpy()
                user_embeddings[user_id] = embeddings
        return user_embeddings

    def evaluate_trials(self, user_embeddings, k_shots):
        genuine_scores, impostor_scores = [], []
        user_ids = list(user_embeddings.keys())

        for uid in user_ids:
            all_seqs = user_embeddings[uid]
            if len(all_seqs) <= k_shots:
                continue

            profile = np.mean(all_seqs[:k_shots], axis=0)
            profile = profile / np.linalg.norm(profile)

            for probe in all_seqs[k_shots:]:
                genuine_scores.append(1.0 - np.dot(profile, probe))

            for other_uid in user_ids:
                if other_uid == uid:
                    continue
                for probe in user_embeddings[other_uid]:
                    impostor_scores.append(1.0 - np.dot(profile, probe))

        return np.array(genuine_scores), np.array(impostor_scores)

    def compute_metrics(self, genuine_scores, impostor_scores):
        if len(genuine_scores) == 0 or len(impostor_scores) == 0:
            return {"EER": 0.0, "ROC-AUC": 0.0, "TAR@1%FAR": 0.0}

        y_true = np.concatenate([np.zeros(len(genuine_scores)), np.ones(len(impostor_scores))])
        y_scores = np.concatenate([genuine_scores, impostor_scores])

        fpr, tpr, thresholds = roc_curve(y_true, -y_scores)
        roc_auc = auc(fpr, tpr)

        try:
            eer = brentq(lambda x: 1.0 - x - interp1d(fpr, tpr)(x), 0.0, 1.0)
        except:
            eer = 0.0

        try:
            tar_at_far = float(interp1d(fpr, tpr, fill_value="extrapolate")(0.01))
        except:
            tar_at_far = 0.0

        return {"EER": float(eer), "ROC-AUC": float(roc_auc), "TAR@1%FAR": tar_at_far}

# ==========================================
# 4. AUTOMATIC DYNAMIC TRAINING & EVALUATION
# ==========================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Automatically detect input dimension from parsed data shape
sample_user_seqs = next(iter(data_by_user.values()))
input_dim = sample_user_seqs[0].shape[1]
print(f"Automatically detected input_dim: {input_dim}")

dataset = BehavioralDataset(data_by_user)
dataloader = DataLoader(dataset, batch_size=32, shuffle=True)

model = TransformerEmbedding(input_dim=input_dim).to(device)

optimizer = optim.Adam(model.parameters(), lr=0.0005, weight_decay=1e-4)
criterion = nn.TripletMarginLoss(margin=0.2, p=2)

epochs = 50
print(f"Starting training for Siamese Transformer over {epochs} epochs...")

model.train()
for epoch in range(epochs):
    total_loss = 0.0
    for anchor, positive, negative in dataloader:
        anchor, positive, negative = anchor.to(device), positive.to(device), negative.to(device)
        optimizer.zero_grad()
        anc_emb, pos_emb, neg_emb = model(anchor, positive, negative)
        loss = criterion(anc_emb, pos_emb, neg_emb)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()

    if (epoch + 1) % 5 == 0 or epoch == 0:
        print(f"Epoch [{epoch+1}/{epochs}], Loss: {total_loss / len(dataloader):.4f}")

# Evaluation Phase
print("\nEvaluating Siamese Transformer Across Few-Shot Protocols:")
evaluator = BiometricEvaluator(model, device)
user_embeddings = evaluator.get_embeddings(data_by_user)

shot_counts = [1, 3, 5, 10]
for k in shot_counts:
    gen_scores, imp_scores = evaluator.evaluate_trials(user_embeddings, k_shots=k)
    metrics = evaluator.compute_metrics(gen_scores, impostor_scores=imp_scores)
    print(f"  -> {k}-shot | EER: {metrics['EER']:.4f} | AUC: {metrics['ROC-AUC']:.4f} | TAR@1%FAR: {metrics['TAR@1%FAR']:.4f}")
```
